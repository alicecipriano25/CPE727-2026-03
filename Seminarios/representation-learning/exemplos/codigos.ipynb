```python
from collections import defaultdict
from statistics import mean, stdev

def resumir(linhas, chave):
    grupos = defaultdict(list)
    for linha in linhas:
        grupos[(linha[chave], linha["metodo"])].append(linha["acuracia"])
    for (cenario, metodo), valores in grupos.items():
        print(f"{cenario!s:10s} {metodo:22s} {mean(valores):.2%} ± {stdev(valores):.2%}")
```
